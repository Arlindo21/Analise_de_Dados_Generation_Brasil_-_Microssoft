```python
#s achatar o json, basta usar a função json_normalize do pandas
import pandas as pd
import requests #para fazer requisições http
url_raw = input('cole o link do json aqui')
resposta = requests.get(url_raw)
#va ate ao git pegar o arqivo json e copie o link do raw[]
dados=resposta.json()
df_json=pd.DataFrame(dados)

#agora vamos arrumar os dados para table 
try:
    nome_tabela = input('digite o nome da tabela que deseja criar')
    tabela_dados = pd.json_normalize(dados[nome_tabela])
    display(tabela_dados)
except Exception as e:
    print(f"Não conssigo ler : {e}")
```

```python
#padronização da informação 

tabela_dados['department.name'] = tabela_dados['department.name'].str.strip().str.upper()

print(tabela_dados['department.name'])


#renomeiar o que dever ser atualizado 
tabela_dados=tabela_dados.rename(columns={'nome':'Nome'
    'department.name':'Nome do Departamento'
    })
```
